# List of models I have access to

```python
import openai

def list_openai_models():
    """
    Lists all models available to the authenticated OpenAI client.
    """
    client = openai.OpenAI()  # If you use an API key, pass it as: openai.OpenAI(api_key="YOUR_KEY")
    models = client.models.list()
    return [model.id for model in models.data]

# Example usage:
available_models = list_openai_models()
print(available_models)
```
